This notebook analyzes a small sales dataset using Python functions. Each function takes the full dataset as its first argument.

```python
sales_data = [
('Laptop', 'East', 1200.00),
('Phone', 'West', 850.00),
('Tablet', 'East', 400.00),
('Monitor', 'South', 650.00),
('Keyboard', 'West', 75.00),
('Laptop', 'South', 1150.00),
('Phone', 'East', 900.00),
('Tablet', 'West', 425.00),
('Monitor', 'East', 680.00),
('Keyboard', 'South', 80.00),
('Laptop', 'West', 1300.00),
('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region): return sum(amt for _, r, amt in data if r == region)
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    #num_of_records = 0
    return sum(1 for _, r, amt in data if r == region)
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
import statistics
def region_average(data, region):
    amounts = [amt for _, r, amt in data if r == region]
    
    if len(amounts) == 0:
        return 0
    
    return statistics.mean(amounts)
```

```python
print(f"${region_average(sales_data, 'East'):.2f}")
print(f"${region_average(sales_data, 'West'):.2f}")
print(f"${region_average(sales_data, 'South'):.2f}")
print(f"${region_average(sales_data, 'North'):.2f}")
```

## Lab 5. Top and bottom sale

```python
def top_sale(sales_data):
    biggest_sale = max(sales_data, key = lambda x: x[2]) 
    return biggest_sale
print(top_sale(sales_data))
```

```python
def bottom_sale(sales_data):
    lowest_sale = min(sales_data, key = lambda x: x[2]) 
    return lowest_sale
print(bottom_sale(sales_data))
```

## Lab 7. Unique regions

```python
## set(["East", 'West', 'East', 'East', 'West'])
def unique_regions(data):
    return sorted(set(i[1] for i in data))
print(unique_regions(sales_data))

```
