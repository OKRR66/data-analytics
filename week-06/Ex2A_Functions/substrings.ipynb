## Lab 1. Working with names - parsing strings

```python
test_name = 'Megan Thee Stallion Camille Jayce Kaitlyn'
```

```python
def parse_and_display(name):
    for x in range(len(name)):
        if name[x] == ' ':
            first = name[:x]
            last = name[x+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
            
```

```python
parse_and_display(test_name)
```

```python
def parse_and_display2(test_name):
    parts = test_name.split()
    if len(parts) == 1:
        print(f"First name is {parts[0]}")
        print("No last name")
    elif len(parts) == 2:
        print(f"First name is {parts[0]}")
        print(f"Last name is {parts[1]}")
    else:
        print(f"First name is {parts[0]}")
        print(f"Middle name is {parts[1:-1]}")
        print(f"Last name is {parts[-1]}")
parse_and_display2(test_name)
```
