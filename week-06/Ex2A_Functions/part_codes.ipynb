## Lab 2. Parsing strings – part codes

```python
## ACME:123-L supplied by Acme, part #123, size large 
## DI:12345-M supplied by DI, part #12345, size medium 
## ACE:1-12 supplied by Ace, part #1, size 12
```

```python
def get_supplier_code(part_code):
    colon = part_code.index(":")
    return part_code[:colon]
```

```python
def get_product_num(part_code):
    colon = part_code.index(":")
    dash = part_code.index("-")
    return part_code[colon+1:dash]
```

```python
def get_size(part_code):
    dash = part_code.index("-")
    return part_code[dash+1:]
```

```python
# Değişkenler — ayrı cell
part1 = "ACME:123-L"
part2 = "DI:12345-M"
part3 = "ACE:1-12"
```

```python
print(f"For {part1}, the supplier is {get_supplier_code(part1)}. Part number {get_product_num(part1)} is available in size {get_size(part1)}.")
```

```python
print(f"For {part2}, the supplier is {get_supplier_code(part2)}. Part number {get_product_num(part2)} is available in size {get_size(part2)}.")
```

```python
print(f"For {part3}, the supplier is {get_supplier_code(part3)}. Part number {get_product_num(part3)} is available in size {get_size(part3)}.")
```

```python
all_products = [part1, part2, part3]
```

```python
print(all_products)
```

```python
for part in all_products:
    print(f"For {get_supplier_code(part)}, the supplier is {get_supplier_code(part)}. Part number {get_product_num(part)} is available in size {get_size(part)}.")
```
