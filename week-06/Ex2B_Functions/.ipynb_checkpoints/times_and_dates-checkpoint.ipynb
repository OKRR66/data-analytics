## working with datetime functions.

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
## print(f"Today is {today.datetime.strftime(%A)} , {today.datetime.strftime(%B)} , {today.datetime.strftime(%D)} , {today.datetime.strftime(%Y)}")

print(f"Today is {today.strftime('%A, %B %d, %Y')}")
```

# My birthday

```python
my_birthday = datetime.datetime(1996, 5, 13)
print(my_birthday)
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
print(f"90 days from now is {ninety_d.strftime("%B , %d, %Y")}")
```

# Time for Dinner

```python
dinner_time = datetime.time(18, 30)
print(dinner_time)
print(f"Let's meet for dinner at {dinner_time.strftime('%H:%M')} PM")
```
