 ## Generating Random Choices from an Array

```python
students = np.array(['Longley', 'Maritza', 'Rodas', 'Jessica', 'Eyerusalem', 'Mina', 'Godknows', 'Jimmy', 'Paul', 'Will', 'Sultan', 'Cagla'])
```

```python
call_on = np.random.choice(students,1)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, (3,4))
print(breakout_groups)
#It wouldn't be a good way to create breakout rooms because the array contains 12 elements and we need to pick 12 random elements.
#It is almost impossible to create a list with unique names. We need to decrease the amount of selection or increase the number of sample amount
#Also same name can be in two different groups this is also not logical for this operation.
```

```python
call_on_weighted = np.random.choice(students, 5, p=[0.3, 0.0, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07])
print(call_on_weighted)
#I gave 30% and %0 possibility in p values as question was asking from me. I just seperated the rest of the 
# 70% of the possibility equally between rest of the 10 students
```
