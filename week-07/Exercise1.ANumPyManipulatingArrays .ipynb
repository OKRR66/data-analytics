 ## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

#joined_array_1 was 4x3 2D array. It is splitted into two equal parts from so became 2x3,2x3 two array appeared
#They both have 2 dimesions.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
#This one also splitted from rows, but this time they are not equal. first array is 2x3 and the other two is 1x3
#They are all still 2D but their shape is different
```
