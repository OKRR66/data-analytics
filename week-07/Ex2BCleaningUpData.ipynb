## Modifying a DataFrame.

```python
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

```python
df_cleaned = df_cleaned.drop(columns=['City'])
df_cleaned
```

```python
""" df_cleaned = df_cleaned['Age'].fillna(df_cleaned['Age'].mean()) """ #If i start the code with df_cleaned instead of
                                                                #df_cleaned['Age'] it deletes rest of the columns
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
df_cleaned
```

```python
df_cleaned['Score'] = df_cleaned['Score'].replace(0, df_cleaned['Score'].mean())
df_cleaned
#There was no 0 score values so df stayed the same.
```

```python
df_cleaned = df_cleaned.dropna()
df_cleaned
```

```python
df_cleaned = df_cleaned.set_index('ID') #This removes index column and sets ID column as Index
df_cleaned
```
