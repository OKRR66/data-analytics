# Exercise 3.A Getting familiar with Matplotlib 


```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5] 
y = [2, 3, 5, 7, 11] 
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```

## Styling Plots

```python
plt.plot(x,y,
         linestyle='--',
         color='red')
```

```python
plt.plot(x,y,
         linestyle='--',
         color='green',
         marker='s',
         markersize=10
         )
```

```python
plt.plot(x,y,
         linestyle='--',
         color='green',
         marker='s',
         markersize=10,
         markerfacecolor='#f99999',
         alpha=0.3
        
         )
```
