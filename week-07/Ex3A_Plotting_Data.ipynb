# Exercise 3.A Getting familiar with Matplotlib 


```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5] 
y = [2, 3, 5, 7, 11] 
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```

## Styling Plots

```python
plt.plot(x,y,
         linestyle='--',
         color='red')
```

```python
plt.plot(x,y,
         linestyle='--',
         color='green',
         marker='s',
         markersize=10
         )
```

```python
plt.plot(x,y,
         linestyle='--',
         color='green',
         marker='s',
         markersize=10,
         markerfacecolor='#f99999',
         alpha=0.3
        
         )
```

# Exercise 3.B More plotting practice 


## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D'] 
sales = [23, 17, 35, 29] 
```

```python
plt.bar(products, sales)
```

```python
plt.bar(products,sales, color='skyblue', width=0.5)
plt.xlabel('Products')
plt.ylabel('Sales')
```

```python
x = np.random.rand(50) 
y = np.random.rand(50)
plt.scatter(x,y)
```

```python
x = np.random.rand(50) 
y = np.random.rand(50)
markersize= np.random.rand(50) * 100
plt.scatter(x,y ,s=markersize, c=y, cmap='hot')
```

 ## Subplots

```python
fig, (ax1, ax2)=plt.subplots(1,2, figsize=(10,5))
x = np.linspace(0, 10, 100) 
y = np.sin(x)
ax1.plot(x,y)
ax1.set_title('Sine Wave')
ax1.set_xlabel('X')
ax1.set_ylabel('sin(x)')

random_data= np.random.randn(1000)
ax2.hist(random_data)
ax2.set_title('Normal Distrubiton')
ax2.set_xlabel('Values')
ax2.set_ylabel('Percent')
plt.tight_layout #tightlayout to prevent writing axis names or numbers on top of eachother
plt.show()
```
