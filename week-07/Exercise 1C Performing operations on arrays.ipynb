## Grocery Problem

```python
bag1 = np.array([2,3,4,0])
bag2 = np.array([4,0,1,2])
print(bag1)
print(bag2)
```

```python
print(f'The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.')
```

```python
totals = np.vstack((bag1,bag2))
print(totals)
totals_cat = np.sum(totals, axis=0)
print(f'There are {totals_cat[0]} vegetables, {totals_cat[1]} fruits, {totals_cat[2]}, boxes of snacks, and {totals_cat[3]} beverages in the two grocery bags.')
```
