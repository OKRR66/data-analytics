# Exercise 2.C Working with multiple DataFrames 


```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'], 
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 
'West', 'North'], 
'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15], 
'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]} 
```

```python
df_sales = pd.DataFrame(data)
df_sales
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
df_sales['Units_Sold'].sum()

```

```python
df_sales['Revenue'].sum()
```

```python
df_sales['Units_Sold'].mean().round(2)
```

```python
df_sales['Revenue'].mean().round(2)
```

## Grouping Data

```python
df_sales.groupby('Product')[['Units_Sold','Revenue']].sum()
```

```python
df_sales.groupby('Region')[['Units_Sold','Revenue']].mean().round(2)
```

```python
df_sales.groupby(['Product','Region'])[['Units_Sold', 'Revenue']].sum()
```

## Sorting and Ranking Data

```python
df_sales.sort_values('Units_Sold',ascending=True)
```

```python
df_sales.sort_values('Revenue',ascending=False)
```

```python
df_sales['Revenue_Rank']= df_sales['Revenue'].rank(ascending=False)
df_sales
#Since there are equal values results come as float. to make them integer: 
#df_sales['Revenue'].rank(ascending=False, method='first')
# method='first' When there are equal values which ever is first gets the lower rank.
```

 ## Merging DataFrames

```python
product_data = {'Product': ['A', 'B', 'C'], 
'Category': ['Electronics', 'Furniture', 'Appliances']} 
df_products = pd.DataFrame(product_data) 
```

```python
df_merged = pd.merge(df_sales , df_products, on='Product')
df_merged
```

```python
df_merged.groupby('Category')['Revenue'].sum()
```
